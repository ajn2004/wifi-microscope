# Wi-Fi Microscope: CSI scientific exploration

**Question:** what information is observable in these two fixed-radio measurements, and does the controlled arm A/B/A perturbation produce a repeatable signature?

Raw files remain untouched. All time axes are capture-relative and independently zeroed per file; receiver timestamps are treated as uint32 microseconds and unwrapped only to diagnose resets. The perturbation boundaries are nominal, not synchronized markers. Coefficient indices below are storage positions, **not** asserted subcarrier frequencies.

The parser follows the record emitted by `firmware/reciever/main/receiver.c`. CSI decoding uses signed int8 `(imaginary, real)` pairs and masks the first four bytes when `first_word_invalid` is set. See the [ESP-IDF v5.5.6 ESP32 CSI guide](https://docs.espressif.com/projects/esp-idf/en/v5.5.6/esp32/api-guides/wifi.html#esp32-wi-fi-channel-state-information) and its CSI configure section for exact layout/index interpretation; training-field concatenation depends on PHY and configuration, so no frequency mapping is assumed here.

Run from repository root or this notebook directory. Dependencies: `pip install -r notebooks/requirements.txt`.


```python
from pathlib import Path
import re, json, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import signal, stats
from sklearn.decomposition import PCA

plt.rcParams.update({"figure.figsize": (10, 4), "axes.grid": True, "grid.alpha": .25})
BASELINE_PATH = Path("../data/raw/csi-baseline-01.log")
PERTURBATION_PATH = Path("../data/raw/csi-perturbation-01.log")
# From notebooks/, paths above resolve. From repository root, use data/raw/... instead.
for name, path in [("baseline", BASELINE_PATH), ("perturbation", PERTURBATION_PATH)]:
    if not path.exists() and Path("data/raw", path.name).exists():
        if name == "baseline": BASELINE_PATH = Path("data/raw", path.name)
        else: PERTURBATION_PATH = Path("data/raw", path.name)
    print(f"{name}: {BASELINE_PATH if name == 'baseline' else PERTURBATION_PATH} — {'present' if (BASELINE_PATH if name == 'baseline' else PERTURBATION_PATH).exists() else 'MISSING; set configurable path above'}")
```

## 1. Acquisition and parsing

Only lines containing a `CSI,` record are considered. ANSI escapes and leading monitor prefixes are tolerated; unrelated boot/serial lines are ignored. A candidate CSI line that fails validation is reported with source line and reason. Any timestamp decrease is interpreted as a uint32 wrap when consistent with wraparound, otherwise a reset/new segment; inter-record intervals remain diagnostic and do not prove RF packet loss.

```python
FIELDS = ["timestamp_us", "rssi_dbm", "channel", "sig_mode", "rate_code", "first_word_invalid", "csi_len", "csi_hex"]
def parse_capture(path):
    rows, issues = [], []
    ansi = re.compile(r"\x1b\[[0-?]*[ -/]*[@-~]")
    for line_no, raw in enumerate(Path(path).read_text(errors="replace").splitlines(), 1):
        line = ansi.sub("", raw)
        match = re.search(r"CSI,(.*)$", line)
        if not match: continue
        fields = match.group(1).strip().split(",")
        try:
            if len(fields) != 8: raise ValueError(f"expected 8 fields, found {len(fields)}")
            row = dict(zip(FIELDS, fields))
            for key in FIELDS[:-1]: row[key] = int(row[key], 10)
            raw_bytes = bytes.fromhex(row["csi_hex"])
            if len(row["csi_hex"]) % 2: raise ValueError("odd-length hex payload")
            if len(raw_bytes) != row["csi_len"]: raise ValueError(f"decoded {len(raw_bytes)} bytes, declared {row['csi_len']}")
            if len(raw_bytes) % 2: raise ValueError("CSI byte count is not a complete I/Q pair")
            row["raw_bytes"] = raw_bytes
            row["source_line"] = line_no
            rows.append(row)
        except Exception as exc:
            issues.append({"line": line_no, "reason": str(exc), "text": raw[:180]})
    if not rows: raise ValueError(f"No valid CSI records in {path}")
    df = pd.DataFrame(rows)
    lengths = df.csi_len.unique()
    if len(lengths) != 1: raise ValueError(f"Variable payload lengths {lengths}; cannot form aligned matrix")
    raw = np.stack([np.frombuffer(b, dtype=np.uint8) for b in df.raw_bytes])
    signed = raw.view(np.int8).astype(float).reshape(len(df), -1, 2)
    # ESP-IDF layout: imaginary byte then real byte.
    H = signed[:, :, 1] + 1j * signed[:, :, 0]
    mask = np.zeros(H.shape[1], dtype=bool)
    if df.first_word_invalid.any(): mask[:min(2, len(mask))] = True
    H[:, mask] = np.nan + 1j*np.nan
    # uint32 wrap vs a genuine timer reset; maintain a monotonic diagnostic timeline by segments.
    ts = df.timestamp_us.to_numpy(dtype=np.int64)
    elapsed = np.zeros(len(ts), dtype=float); segment = np.zeros(len(ts), dtype=int); offset = 0; seg = 0
    for i in range(1, len(ts)):
        delta = int(ts[i] - ts[i-1])
        if delta < 0:
            if ts[i-1] > 0xF0000000 and ts[i] < 0x10000000:
                offset += 2**32
            else:
                seg += 1; offset = int(elapsed[i-1]*1e6) - int(ts[i])
        elapsed[i] = (ts[i] + offset)/1e6
        segment[i] = seg
    df["capture_time_s"] = elapsed - elapsed[0]
    df["timestamp_segment"] = segment
    dt = np.diff(df.capture_time_s)
    df["interval_s"] = np.r_[np.nan, dt]
    return df, H, mask, pd.DataFrame(issues)

captures = {}
for key, path in [("baseline", BASELINE_PATH), ("perturbation", PERTURBATION_PATH)]:
    if path.exists():
        captures[key] = parse_capture(path)
        print(f"{key}: {len(captures[key][0])} valid records; {len(captures[key][3])} malformed CSI candidates")
        display(captures[key][3].head(10))
    else: print(f"{key}: unavailable; analyses requiring it will be skipped")
```

## 2. Data-quality assessment

Intervals reflect observed receiver record timestamps. Long intervals can arise from RF arrival cadence, callback/queue/serial buffering, or logging loss; no sender sequence number or host-arrival timestamps are present here, so they are not labeled packet loss. Int8 endpoint counts are a clipping indicator, not proof of analog saturation.

```python
def quality_report(df, H, mask, issues, label):
    dt = df.interval_s.dropna()
    mag = np.abs(H)
    byte_matrix = np.stack([np.frombuffer(b, dtype=np.uint8) for b in df.raw_bytes])
    endpoint_fraction = np.mean(np.isin(byte_matrix, [0x7f, 0x80]))
    print(f"{label}: n={len(df)}, duration={df.capture_time_s.iloc[-1]:.3f} s, malformed={len(issues)}")
    print(f"interval median={dt.median():.4f} s, 95%={dt.quantile(.95):.4f} s, max={dt.max():.4f} s; >2x median={(dt > 2*dt.median()).sum()}")
    print(f"RSSI median={df.rssi_dbm.median():.1f} dBm, range=[{df.rssi_dbm.min()}, {df.rssi_dbm.max()}], SD={df.rssi_dbm.std():.2f} dB")
    print(f"payload lengths={df.csi_len.value_counts().to_dict()}, sig_mode={df.sig_mode.value_counts().to_dict()}, rate_code={df.rate_code.value_counts().to_dict()}, channel={df.channel.value_counts().to_dict()}")
    print(f"masked coefficient indices={np.flatnonzero(mask).tolist()}; always-zero unmasked indices={np.flatnonzero(np.all(np.nan_to_num(H)==0, axis=0) & ~mask).tolist()}")
    print(f"CSI magnitude endpoints (int8 -128/+127) fraction={endpoint_fraction:.5f}; zero raw-byte fraction={np.mean(byte_matrix == 0):.3f}")
    print(f"timestamp segments={df.timestamp_segment.nunique()}; no sender sequence/host-arrival/drop counters, so RF loss vs buffering is unresolved")
    fig, ax = plt.subplots(2,1, sharex=True, figsize=(10,5))
    ax[0].plot(df.capture_time_s, df.rssi_dbm, '.', ms=2); ax[0].set_ylabel("RSSI (dBm)"); ax[0].set_title(label)
    ax[1].plot(df.capture_time_s, df.interval_s, '.', ms=2); ax[1].axhline(dt.median(), color='C1', label='median'); ax[1].set_ylabel("Record interval (s)"); ax[1].set_xlabel("Capture-relative time (s)"); ax[1].legend(); plt.tight_layout()
for label, (df,H,mask,issues) in captures.items(): quality_report(df,H,mask,issues,label)
```

## 3. Stationary baseline

Amplitude is `|H|` in arbitrary receiver units; phase is `arg(H)` in radians. Packet normalization divides by the RMS over valid coefficients, removing scalar amplitude but not phase. The nuisance model is fitted packetwise by complex least squares: for template `H0`, `c(t)=<H0,H(t)>/<H0,H0>`, so `a=|c|`, `phi=arg(c)` and residual `H-cH0`. Explained variation is computed relative to the grand-mean template and compared with a centered variance baseline.

```python
if "baseline" in captures:
    df,H,mask,_ = captures["baseline"]; valid = ~mask & np.isfinite(H).all(axis=0)
    A = np.abs(H[:,valid]); phase = np.angle(H[:,valid]); t=df.capture_time_s.to_numpy(); idx=np.flatnonzero(valid)
    fig,ax=plt.subplots(3,1,figsize=(11,8),sharex=True)
    ax[0].plot(t,df.rssi_dbm); ax[0].set_ylabel("RSSI (dBm)")
    ax[1].plot(idx,A.mean(axis=0)); ax[1].set_ylabel("Mean |CSI| (a.u.)")
    ax[2].plot(idx,np.nanmedian(np.unwrap(phase,axis=1),axis=0)); ax[2].set_ylabel("Unwrapped phase (rad)"); ax[2].set_xlabel("Opaque coefficient index")
    plt.tight_layout()
    rms=np.sqrt(np.mean(A**2,axis=1)); normalized=A/rms[:,None]
    fig,ax=plt.subplots(1,2,figsize=(12,4)); ax[0].plot(t,rms); ax[0].set(ylabel="Packet RMS |CSI| (a.u.)",xlabel="Capture-relative time (s)")
    im=ax[1].imshow(A.T,aspect='auto',origin='lower',extent=[t[0],t[-1],idx[0],idx[-1]]); ax[1].set(xlabel="Capture-relative time (s)",ylabel="Opaque coefficient index",title="Raw CSI amplitude"); fig.colorbar(im,ax=ax[1],label="|CSI| (a.u.)"); plt.tight_layout()
    stats_table=pd.DataFrame({"mean":np.nanmean(A,axis=0),"variance":np.nanvar(A,axis=0,ddof=1),"median":np.nanmedian(A,axis=0),"CV":np.divide(np.nanstd(A,axis=0,ddof=1),np.nanmean(A,axis=0),out=np.full(A.shape[1],np.nan),where=np.nanmean(A,axis=0)!=0)},index=idx)
    display(stats_table.describe())
    variable=np.nanstd(A,axis=0)>0; corr=np.corrcoef(A[:,variable],rowvar=False); print(f"Amplitude correlations computed for {variable.sum()} nonconstant coefficients; constants omitted."); plt.figure(figsize=(7,6)); plt.imshow(corr,vmin=-1,vmax=1,cmap='coolwarm',origin='lower'); plt.colorbar(label='Amplitude correlation'); plt.title("Inter-coefficient amplitude correlation"); plt.xlabel("Nonconstant coefficient ordinal"); plt.ylabel("Nonconstant coefficient ordinal")
    fig,ax=plt.subplots(figsize=(10,4)); ax.plot(t,A[:,0],alpha=.7,label='raw coefficient 0'); ax.plot(t,normalized[:,0],alpha=.7,label='packet-RMS normalized'); ax.set(xlabel="Time (s)",ylabel="Amplitude (a.u.)",title="Raw vs packet-amplitude normalized"); ax.legend()
    # Scalar complex nuisance fit to the mean CSI template.
    h0=np.nanmean(H,axis=0); h=H[:,valid]; template=h0[valid]
    c=np.einsum('k,tk->t',np.conj(template),h)/np.vdot(template,template)
    fit=c[:,None]*template[None,:]; residual=h-fit
    total=np.sum(np.abs(h-template[None,:])**2); residual_ss=np.sum(np.abs(residual)**2)
    print(f"Scalar complex-template model: residual SSE / grand-template SSE={residual_ss/total:.3f}; variance explained={1-residual_ss/total:.3f}; mean |c|={np.mean(np.abs(c)):.3g}, SD |c|={np.std(np.abs(c)):.3g}")
    fig,ax=plt.subplots(1,2,figsize=(11,4)); ax[0].plot(t,np.abs(c),label='a(t)'); ax[0].plot(t,np.unwrap(np.angle(c)),label='phi(t) (rad)'); ax[0].legend(); ax[0].set_xlabel("Time (s)")
    ax[1].imshow(np.abs(residual).T,aspect='auto',origin='lower',extent=[t[0],t[-1],idx[0],idx[-1]]); ax[1].set(title="Magnitude of complex-fit residual",xlabel="Time (s)",ylabel="Opaque coefficient index")
    # Temporal autocorrelation and Welch spectrum, using regular-grid interpolation only if cadence reasonably stable.
    y=rms-np.mean(rms); med=np.median(np.diff(t)); regular=np.max(np.abs(np.diff(t)-med)) < .2*med
    if regular and med>0:
        ac=signal.correlate(y,y,mode='full',method='fft')[len(y)-1:]; ac=ac/ac[0]
        f,p=signal.welch(y,fs=1/med,nperseg=min(256,len(y)))
        fig,ax=plt.subplots(1,2,figsize=(10,3)); ax[0].plot(np.arange(min(100,len(ac)))*med,ac[:100]); ax[0].set(xlabel="Lag (s)",ylabel="Autocorrelation")
        ax[1].semilogy(f,p); ax[1].set(xlabel="Frequency (Hz)",ylabel="PSD (a.u.²/Hz)"); plt.tight_layout()
    else: print("Autocorrelation/spectrum omitted: sampling is too irregular for an unqualified uniform-grid estimate.")
```

## 4. A/B/A perturbation

Nominal phases use 0–30 s (A1), 30–60 s (B), and 60–90 s (A2). A fixed ±2 s exclusion around nominal boundaries is used for steady-state comparisons; a transparent moving-change diagnostic below reports candidate change times rather than silently shifting labels. Bootstrap intervals resample contiguous blocks (2 s) within each phase; these quantify within-run sampling variability, not between-session reproducibility.

```python
if "perturbation" in captures:
    df,H,mask,_=captures["perturbation"]; t=df.capture_time_s.to_numpy(); valid=~mask & np.isfinite(H).all(axis=0); A=np.abs(H[:,valid]); rms=np.sqrt(np.mean(A**2,axis=1)); norm=A/rms[:,None]
    cond=np.where(t<30,"A1",np.where(t<60,"B","A2")); stable=(t>2)&(t<28) | (t>32)&(t<58) | (t>62)&(t<88)
    fig,ax=plt.subplots(3,1,figsize=(11,9),sharex=True)
    ax[0].plot(t,df.rssi_dbm,'.',ms=2); ax[0].set_ylabel("RSSI (dBm)")
    for x in [30,60]: ax[0].axvline(x,color='k',ls='--')
    for vals,title,a in [(A,"Amplitude |CSI|",ax[1]),(norm,"Packet-RMS normalized |CSI|",ax[2])]:
        a.imshow(vals.T,aspect='auto',origin='lower',extent=[t[0],t[-1],0,vals.shape[1]-1],cmap='viridis'); a.set_ylabel("Valid coefficient ordinal"); a.set_title(title)
        for x in [30,60]: a.axvline(x,color='white',ls='--')
    ax[2].set_xlabel("Capture-relative time (s)"); plt.tight_layout()
    # Profiles and block bootstrap confidence intervals for each segment.
    rng=np.random.default_rng(20251030); block=max(1,int(round(2/np.median(np.diff(t))))); boot={}
    fig,ax=plt.subplots(figsize=(10,4))
    for label,(lo,hi) in {"A1":(2,28),"B":(32,58),"A2":(62,88)}.items():
        sel=(t>=lo)&(t<hi); vals=A[sel]; n=len(vals); means=vals.mean(axis=0)
        samples=[]
        for _ in range(500):
            starts=rng.integers(0,max(1,n-block+1),size=int(np.ceil(n/block)))
            take=np.concatenate([np.arange(s,min(s+block,n)) for s in starts])[:n]
            samples.append(vals[take].mean(axis=0))
        low,high=np.percentile(samples,[2.5,97.5],axis=0); boot[label]=(means,low,high)
        x=np.arange(A.shape[1]); ax.plot(x,means,label=label); ax.fill_between(x,low,high,alpha=.18)
    ax.set(xlabel="Valid coefficient ordinal (not frequency)",ylabel="Mean |CSI| (a.u.)",title="Steady-state amplitude profiles ± 95% block-bootstrap CI"); ax.legend(); plt.tight_layout()
    # Per-feature Euclidean profile distances, and global scalar effect distances.
    means={k:v[0] for k,v in boot.items()}
    print("Profile RMS distances (a.u.):",{f"{x}-{y}":float(np.sqrt(np.mean((means[x]-means[y])**2))) for x,y in [("A1","A2"),("A1","B"),("A2","B")]})
    print("Overall amplitude RMS by condition:",{k:float(np.mean(rms[(cond==k)&stable])) for k in ["A1","B","A2"]})
    # PCA/SVD on normalized amplitude, centered over steady-state observations.
    X=norm[stable]; labels=cond[stable]; pca=PCA(n_components=min(5,X.shape[0],X.shape[1])).fit(X); scores=pca.transform(X)
    print("PCA explained variance ratio:",np.round(pca.explained_variance_ratio_,4))
    fig,ax=plt.subplots(2,1,figsize=(10,6),sharex=True); ax[0].plot(t,rms); ax[0].set_ylabel("Packet RMS (a.u.)")
    ax[1].scatter(t[stable],scores[:,0],c=pd.Categorical(labels,categories=["A1","B","A2"]).codes,cmap='viridis',s=8); ax[1].set(xlabel="Time (s)",ylabel="PC1 score",title="Normalized-amplitude PCA score; color codes A1/B/A2"); plt.tight_layout()
    for j in range(min(3,scores.shape[1])):
        print(f"PC{j+1} condition means:",{k:float(scores[labels==k,j].mean()) for k in ["A1","B","A2"]})
    # Coarse change-point screen: rolling median RMS, largest changes in 2 s binned series.
    bin_s=2.; edges=np.arange(0,t[-1]+bin_s,bin_s); binned=pd.Series(rms,index=t).groupby(pd.cut(t,edges,include_lowest=True)).median().dropna()
    changes=np.diff(binned.to_numpy()); top=np.argsort(np.abs(changes))[-4:]
    print("Largest adjacent 2-s median RMS changes near (s):",[float(edges[min(i+1,len(edges)-1)]) for i in top])
```

## 5. Complex CSI and nuisance phase

Complex inner product uses `rho(t,u)=|H(t)^H H(u)|/(||H(t)|| ||H(u)||)`, invariant to a global phase rotation. Packet alignment rotates each packet by `exp(-i arg(H0ᴴH(t)))`; relative coefficient phase is `arg(H_k H_ref*)`, which cancels common phase but remains sensitive to reference nulls. These operations do **not** recover absolute propagation phase or calibrated distance.

```python
def complex_features(H, valid):
    z=H[:,valid]; amp=np.abs(z); ref=z[:,0]
    relative=np.angle(z*np.conj(ref[:,None]))
    h0=np.nanmean(z,axis=0); c=np.einsum('k,tk->t',np.conj(h0),z)/np.vdot(h0,h0)
    aligned=z*np.exp(-1j*np.angle(c))[:,None]
    return z,amp,relative,c,aligned
for key in captures:
    df,H,mask,_=captures[key]; valid=~mask & np.isfinite(H).all(axis=0)
    z,amp,rel,c,aligned=complex_features(H,valid)
    print(key, "median pairwise complex normalized inner product after global-phase invariance:", float(np.median(np.abs(np.sum(np.conj(z[:-1])*z[1:],axis=1))/(np.linalg.norm(z[:-1],axis=1)*np.linalg.norm(z[1:],axis=1)))))
    fig,ax=plt.subplots(1,2,figsize=(11,4)); ax[0].imshow(rel.T,aspect='auto',origin='lower',cmap='twilight',vmin=-np.pi,vmax=np.pi); ax[0].set(title=f"{key}: relative phase to first valid coefficient",xlabel="Packet index",ylabel="Valid coefficient ordinal");
    ax[1].plot(df.capture_time_s,np.unwrap(np.angle(c)),label="common phase φ(t)"); ax[1].set(xlabel="Capture-relative time (s)",ylabel="Aligned common phase (rad)"); ax[1].legend(); plt.tight_layout()
    if key=="perturbation":
        t=df.capture_time_s.to_numpy(); groups=[(t>2)&(t<28),(t>32)&(t<58),(t>62)&(t<88)]
        # Complex discriminability using phase-aligned amplitude and relative-phase feature vector.
        feat=np.concatenate([amp/np.linalg.norm(amp,axis=1)[:,None], np.cos(rel), np.sin(rel)],axis=1)
        names=["A1","B","A2"]; fmeans=[feat[g].mean(axis=0) for g in groups]
        print("Complex-feature RMS distances:",{f"{names[i]}-{names[j]}":float(np.sqrt(np.mean((fmeans[i]-fmeans[j])**2))) for i,j in [(0,2),(0,1),(1,2)]})
        # Compare normalized amplitude-only versus complex feature separation descriptively; dimensions differ, so no direct superiority claim.
```

## 6. Physical interpretation

- **Received-power sensing:** RSSI is a coarse scalar estimate in dBm; it mixes propagation, antenna/radio response, AGC and quantization.
- **Frequency-dependent CSI sensing:** each stored complex slot is a channel estimate in the driver’s returned training-field ordering. Its pattern can reveal differential channel changes across the captured coefficients, but the current parser does not establish a one-to-one physical frequency map.
- **Temporal change detection:** repeated packets at one geometry can detect changes in the measured channel over time, conditional on stability of radios, clocks, environment and acquisition path.
- **Spatial localization:** one Tx–Rx geometry provides no direct independent spatial coordinate. A change may be detectable without determining where it occurred.
- **Inverse-scattering tomography:** reconstructing an image requires a forward model and enough independent, calibrated measurements (e.g., multiple Tx/Rx positions or angles, known geometry/materials, frequency diversity, phase reference and repeatable boundary conditions). A single fixed link over one narrow 20 MHz channel is severely underdetermined.

The arm is an uncontrolled distributed scatterer with uncertain placement/pose and may change coupling or blockage. **No anatomical image or absolute propagation phase is inferred.**
